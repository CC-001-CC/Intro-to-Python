# 1. Defining and calling a function

`def` creates a named block of reusable code.


````python
def greet():
    print("Welcome!")

greet()
````

The two important parts are:

```text
def greet():
    print("Welcome!")
```

This is the **function definition**.

```text
greet()
```

This is the **function call**.

The surrounding code that makes a function call is the **caller**.

When Python reads the definition, the function becomes available. The indented body runs when Python reaches the call.

```text
definition
    ↓
function becomes available

greet()
    ↓
enter function body
    ↓
print("Welcome!")
    ↓
return to caller
```

Here, “return to caller” means execution continues after the call. Returned data values come later in the lesson.

# 2. Parameters and arguments

The first function always prints the same message.

A **parameter** lets a function receive a value from the caller.


````python
def greet(name):
    print(f"Welcome, {name}!")

greet("Maya")
````

In the definition:

```python
def greet(name):
```

`name` is a **parameter**.

In the call:

```python
greet("Maya")
```

`"Maya"` is an **argument**.

```text
argument value
   "Maya"
      ↓
parameter
    name
      ↓
function body
```

The parameter receives the argument value for that call.

# 3. Returning a value

A function that only prints a result can show it to the user.

Other code may also need to **use** the result.

`return` sends a value back to the caller.


````python
def calculate_total(price, quantity):
    total = price * quantity
    return total

cost = calculate_total(12, 3)
print(cost)
````

Trace the returned value:

```text
calculate_total(12, 3)
          ↓
price = 12
quantity = 3
total = 36
          ↓
return 36
          ↓
cost = 36
```

The returned value becomes ordinary data in the caller.


````python
cost = calculate_total(12, 3)
final_cost = cost + 4
print(final_cost)
````

The first call returns `36`. The caller stores that value in `cost`, uses it in another calculation, and prints `40`.


## `print()` versus `return`

These operations have different jobs.

| Operation | Main job |
|---|---|
| `print(value)` | display a value as output |
| `return value` | send a value back to the caller |

Compare these functions:


````python
def show_total(price, quantity):
    print(price * quantity)


def calculate_total(price, quantity):
    return price * quantity
````

### READ

Which function sends a value back to the caller?

Now compare these calls:

```python
show_total(12, 3)
```

and:

```python
cost = calculate_total(12, 3)
```

The first call creates visible output. The second call gives the caller a value it can store and use.

# 4. Local variables and basic scope

Parameters and variables created inside a function belong to that function call.

They are **local** to the function.


````python
def calculate_total(price, quantity):
    total = price * quantity
    return total

cost = calculate_total(12, 3)
print(cost)
````

A useful call-frame model is:

```text
CALL: calculate_total(12, 3)

┌────────────────────────────┐
│ price    ─▶ 12             │
│ quantity ─▶ 3              │
│ total    ─▶ 36             │
└────────────────────────────┘
             │
             └── return 36

caller:
cost ─▶ 36
```

The names `price`, `quantity`, and `total` belong to the function call.

The caller receives the returned value and stores it in `cost`.


### READ

Why does the caller use `cost` after the function finishes?

Consider:

```python
# print(total)
```

The name `total` was created inside `calculate_total`. The caller uses the value that came back through `return`.


# 5. Combining functions

A returned value from one function can become input to another function.


````python
def add_bonus(score, bonus):
    return score + bonus


def subtract_penalty(score, penalty):
    return score - penalty


bonus_score = add_bonus(70, 5)
final_score = subtract_penalty(bonus_score, 2)

print(final_score)
````

Trace the value flow:

```text
70, 5
  ↓
add_bonus
  ↓
75
  ↓
subtract_penalty(..., 2)
  ↓
73
```

The intermediate variable `bonus_score` makes the value flow easy to see.


After the flow is clear, the same operations can be written more compactly:

```python
final_score = subtract_penalty(add_bonus(70, 5), 2)
```

Read from the inner call outward:

1. `add_bonus(70, 5)` returns `75`.
2. `subtract_penalty(75, 2)` returns `73`.

This is similar to the familiar pattern:

```python
int(input("Age: "))
```


# 6. Functions that answer a question

A function can return a Boolean value.


````python
def is_passing(score):
    return score >= 60

print(is_passing(72))
````

`is_passing(72)` returns `True`.

A function that returns `True` or `False` can represent a question about a value.

A Boolean-returning question function is often called a **predicate**.


````python
score = 55

if is_passing(score):
    print("Passing")
else:
    print("Keep working")
````

The call:

```python
is_passing(score)
```

produces a Boolean value that `if` can use as its condition.

# 7. Side effects and `None`

Some functions are mainly useful because of an observable effect.

Displaying output is one example of a **side effect**.


````python
def show_score(score):
    print(f"Score: {score}")

show_score(90)
````

`show_score(90)` displays output.

Now store the result of the call:


````python
def show_score(score):
    print(f"Score: {score}")

result = show_score(90)
print(result)
````

The output is:

```text
Score: 90
None
```

When a function reaches the end without an explicit `return`, Python produces the special value `None`.

Compare the main jobs:

```python
def add_bonus(score, bonus):
    return score + bonus
```

This function is return-focused: it produces a value for the caller.

```python
def show_score(score):
    print(f"Score: {score}")
```

This function is side-effect-focused: it displays output.

Some real functions can do both. This distinction is a useful beginner model for reasoning about function behavior.

# Vocabulary

**function** — a named block of reusable code  
**function definition** — the `def` statement that creates a function  
**function call** — an instruction that runs a function  
**parameter** — a local name in a function definition that receives an argument value  
**argument** — a value supplied in a function call  
**return value** — a value sent back to the caller  
**local variable** — a variable that belongs to a function call  
**scope** — where a name can be used; this lesson focuses on function-local scope  
**side effect** — an observable effect such as displaying output  
**None** — the special value produced when a function finishes without an explicit return
