# 2. Values, types, and variables

A **value** is a piece of data a program can work with.

Two common types of values are:

- a **string**, which represents text,
- an **integer**, which represents a whole number.

Consider these two values:


````python
print("42")
print(42)
````

Both display as `42`, but their types are different.

- `"42"` is a string because it is text inside quotation marks.
- `42` is an integer because it is a whole-number value.

That difference matters when Python performs operations on the values.

## Variables

A **variable** gives a value a name we can use later.

```python
score = 10
```

Read the statement as:

> Assign the value `10` to `score`.

A useful visual model is:

```text
score ─▶ 10
```


````python
score = 10
print(score)
````

# 3. Calculations and updating variables

Python can evaluate arithmetic expressions with numeric values.

For example:

```text
20 + 10 → 30
```

A variable can also be assigned a new value while the program runs.


````python
points = 20
points = points + 10
print(points)
````

Trace the second line in three steps:

```text
points ─▶ 20

READ the current value:
points + 10

CALCULATE:
20 + 10 → 30

ASSIGN the result:
points ─▶ 30
```

Python evaluates the expression on the right first. The result becomes the new value associated with `points`.

# 4. Getting input from a user

Until now, the programmer selected every value.

`input()` lets the person using the program provide text while the program is running.


````python
name = input("What is your name? ")
print(name)
````

The execution flow is:

```text
show prompt
    ↓
user types text
    ↓
input() produces a string
    ↓
name refers to that string
```

The same program can behave differently because the user can enter a different value.

## Putting a variable inside text

An **f-string** lets Python place a variable value inside text.


````python
name = input("What is your name? ")
print(f"Welcome, {name}!")
````

The `f` before the opening quotation mark activates the braces. `{name}` is replaced with the current value of `name`.

## Input produces text

`input()` produces a string, even when the user types digits.

Run the next cell once. Then remove the `#` from the last line and run it again.


````python
age = input("How old are you? ")

# Remove the # below, then run the cell.
# print(age + 1)
````

If you enter `18`, `input()` produces the string:

```text
"18"
```

Python cannot add the string `"18"` and the integer `1`.

The program needs the integer:

```text
18
```

`int` is short for **integer**. `int()` converts numeric text into an integer.


````python
age = input("How old are you? ")
age = int(age)
print(age + 1)
````

Trace the conversion:

```text
"18"
  ↓ int(...)
18
  ↓ + 1
19
```

You may also see the two familiar operations written together:

```python
age = int(input("How old are you? "))
```

Read from the inside outward:

1. `input()` receives text.
2. `int()` converts the numeric text to an integer.
3. the result is assigned to `age`.


# 5. Comparisons and Boolean results

Programs often need to ask questions about values.

A **comparison** asks a yes/no question. Python represents the answer with the Boolean values `True` and `False`.

Before running the next cell, predict the result.


````python
age = 20
print(age >= 18)
````


# 6. Making decisions

A comparison becomes especially useful when Python uses the result to decide which code should run.

The comparison after `if` is called the **condition**.


````python
age = 20

if age >= 18:
    print("You can enter!")
````

Trace the decision:

```text
age >= 18
    ↓
  True
    ↓
run the indented line
```

## Indentation

The indented line belongs to the `if`.

```python
if age >= 18:
    print("You can enter!")
```

Indentation shows which statements belong inside a control structure.

## Adding an `else` path

`else` provides the path used when the `if` condition is `False`.


````python
age = 15

if age >= 18:
    print("You can enter!")
else:
    print("Not yet!")
````

Only one of those two paths runs.

```text
condition
   ↓
 True  ──▶ first path
 False ──▶ else path
```

### MODIFY

Try `age = 20` and `age = 15`. Before each run, predict the path Python will choose.


## More than two paths with `elif`

`elif` means **else if**. It checks another condition after the earlier condition is `False`.


````python
temperature = 70

if temperature > 80:
    print("Hot")
elif temperature > 60:
    print("Comfortable")
else:
    print("Cold")
````

Try the values `90`, `70`, and `40`.

For each value, identify the first condition that becomes `True`. Python runs that path and then continues after the decision structure.

# Vocabulary

**program** — a sequence of instructions for a computer to execute  
**value** — a piece of data a program can work with  
**string** — a text value  
**integer** — a whole-number value  
**variable** — a name used to refer to a value  
**assign** — associate a value with a variable name  
**input** — information received by a program  
**output** — information produced or displayed by a program  
**comparison** — a yes/no question about values  
**condition** — the comparison checked by a decision  
**indentation** — leading spaces that show which statements belong to a block  
**debugging** — finding and fixing a problem in code
