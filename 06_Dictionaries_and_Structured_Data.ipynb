# 2. Adding and updating fields

Dictionaries are mutable.

Assigning to an existing key updates that field.


````python
student = {
    "name": "Maya",
    "score": 92
}

student["score"] = 95

print(student["score"])
````

A new key assignment adds a new field:


````python
student = {
    "name": "Maya",
    "score": 95
}

student["status"] = "ready"

print(student)
````

State model:

```text
before
"name"  ─▶ "Maya"
"score" ─▶ 92

after
"name"   ─▶ "Maya"
"score"  ─▶ 95
"status" ─▶ "ready"
```

This reuses the mutation model from Lesson 5: one existing mutable object changes state.


## Key membership

For a dictionary, `in` checks keys.


````python
student = {
    "name": "Maya",
    "score": 95
}

print("score" in student)
print("email" in student)
````

Expected:

```text
True
False
```

The membership question is:

```text
Does this key exist?
```


# 3. Reading optional fields with `.get()`

Bracket lookup works well when a field is expected to exist:

```python
student["name"]
```

When a field may be absent, `.get()` can provide a fallback.


````python
student = {
    "name": "Maya",
    "score": 95
}

email = student.get("email", "Not provided")

print(email)
````

Expected:

```text
Not provided
```

Mental model:

```text
get(key, fallback)
        ↓
key exists?
├─ yes → stored value
└─ no  → fallback
```

`get()` reads the dictionary. It does not add the missing key.

# 4. Tuples and unpacking

A **tuple** is a small ordered sequence whose positions are fixed.

This lesson uses tuples mainly to prepare for dictionary key-value pairs.


````python
record = ("Maya", 92)

print(record[0])
print(record[1])
````

A tuple can be **unpacked** into multiple names:


````python
record = ("Maya", 92)

name, score = record

print(name)
print(score)
````

Visual:

```text
("Maya", 92)
    ↓      ↓
  name   score
```

**Unpacking** assigns sequence values to multiple names in one step.


# 5. Looping through dictionaries

We can ask a dictionary for its keys, values, or key-value pairs.


## Keys


````python
student = {
    "name": "Maya",
    "major": "Computer Science",
    "score": 92
}

for key in student.keys():
    print(key)
````

## Values


````python
student = {
    "name": "Maya",
    "major": "Computer Science",
    "score": 92
}

for value in student.values():
    print(value)
````

## Key-value pairs

`.items()` provides key-value pairs.


````python
student = {
    "name": "Maya",
    "major": "Computer Science",
    "score": 92
}

for key, value in student.items():
    print(key, value)
````

Connect `.items()` to unpacking:

```text
one pair:
("name", "Maya")
    ↓        ↓
   key     value
```

The loop receives one pair at a time and unpacks it into two names.

# 6. Lists of dictionaries

One dictionary can represent one structured record.

A list can hold several records.


````python
students = [
    {"name": "Maya", "score": 92},
    {"name": "Jordan", "score": 78},
    {"name": "Avery", "score": 85}
]

print(students)
````

Mental model:

```text
students
   ↓
[
  record 0 ─▶ {"name":"Maya",   "score":92}
  record 1 ─▶ {"name":"Jordan", "score":78}
  record 2 ─▶ {"name":"Avery",  "score":85}
]
```

The outer list provides the sequence.

Each dictionary provides named fields.


````python
students = [
    {"name": "Maya", "score": 92},
    {"name": "Jordan", "score": 78},
    {"name": "Avery", "score": 85}
]

for student in students:
    print(student["name"], student["score"])
````

Trace:

| Iteration | Name | Score |
|---:|---|---:|
| 1 | Maya | 92 |
| 2 | Jordan | 78 |
| 3 | Avery | 85 |

The loop variable `student` refers to one dictionary record at a time.


## Mutating records in the list

Because dictionaries are mutable, the loop can update each record.


````python
students = [
    {"name": "Maya", "score": 92},
    {"name": "Jordan", "score": 78},
    {"name": "Avery", "score": 85}
]

for student in students:
    if student["score"] >= 80:
        student["status"] = "ready"
    else:
        student["status"] = "review"

print(students)
````

This reuses the Lesson 5 reference model:

```text
student loop variable
        ↓
dictionary record
        ↓
key assignment changes that record
```

The mutation remains visible through the outer `students` list.


# 7. Sets: uniqueness and membership

A **set** is a collection of unique values.


````python
skills = {"Python", "SQL"}

print("Python" in skills)
````

Repeated equal values do not create repeated entries:


````python
skills = {"Python", "SQL", "Python"}

print(len(skills))
````

Expected length:

```text
2
```

This lesson uses sets only for:

```text
unique values
+
membership
```

Do not depend on a set's printed order.


# 8. Choosing a collection

Different structures express different relationships.

| Need | Collection |

# Debugging: dictionary keys carry meaning

Consider:


````python
student = {
    "name": "Maya",
    "score": 92
}

# A dictionary uses its keys, not list positions:
# print(student[0])

print(student["name"])
````



# Vocabulary

**dictionary** — a mapping from keys to values  
**key** — an identifier for a dictionary field  
**value** — data associated with a key  
**mapping** — a relationship from keys to values  
**field** — a named piece of data in a record  
**record** — related fields describing one entity  
**tuple** — a small ordered sequence with fixed positions  
**unpacking** — assigning sequence values to multiple names  
**set** — a collection of unique values  
**fallback** — a value used when requested data is absent
