# 1. Indexing and item assignment

````python
scores = [88, 91, 76]

print(scores[0])
print(scores[2])
````

Expected output:

```text

```

Now change one position:


````python
scores = [88, 91, 76]

scores[2] = 80

print(scores)
````

```text
before

                   ↑
                 index 2

scores[2] = 80

after
scores ─▶ [88, 91, 80]
```

**Item assignment** replaces the value stored at one list index.

# 3. Mutation

You have already changed a list element.

**Mutation** means changing an existing object in place.


````python
scores = [88, 91, 76]
scores[2] = 80

print(scores)
````

The same list changes state:

```text
[88, 91, 76]
      ↓ change index 2
[88, 91, 80]
```

Lesson 4 gave us an important comparison:

```text
string operation
→ produces a new string
→ string is immutable

list item assignment
→ changes the existing list
→ list is mutable
```


## Mutation and reassignment

These statements have different effects:

```python
scores[2] = 80
```

changes the current list.

```python
scores = [88, 91, 80]
```

assigns the name `scores` to the list created by the new literal.

The reference consequences become important later in this lesson.


# 4. List methods

List methods use the same `value.method()` syntax introduced with strings.


## `append()`

`append(value)` adds one value to the end of the existing list.


````python
scores = [88, 91]
scores.append(76)

print(scores)
````

## `remove()`

`remove(value)` removes the first matching value.


````python
scores = [88, 91, 76]
scores.remove(91)

print(scores)
````

## `pop()`

`pop()` removes the final value and returns it.


````python
scores = [88, 91, 76]
removed = scores.pop()

print(removed)
print(scores)
````

The output is:

```text
76
[88, 91]
```

`pop()` both changes the list and sends the removed value back to the caller.


## `append()` and `None`

Lesson 3 separated returned values from side effects.

Store the result of `append()`:


````python
scores = [88, 91]

result = scores.append(76)

print(scores)
print(result)
````

Expected output:

```text
[88, 91, 76]
None
```

`append()` changes the list as its main effect. It does not return the updated list.

# 5. Building a new list

A program can start with an empty list and append selected values while it processes data.


````python
scores = [72, 91, 84, 65]
high_scores = []

for score in scores:
    if score >= 80:
        high_scores.append(score)

print(high_scores)
````

Trace the new list:

```text
start
[]

read 72 → skip
[]

read 91 → append
[91]

read 84 → append
[91, 84]

read 65 → skip
[91, 84]
```

The reusable pattern is:

```text
start empty
   ↓
loop through source values
   ↓
test/select
   ↓
append selected values
   ↓
new list grows
```

# 6. Lists passed to functions

Function parameters can refer to lists.

Consider:


````python
def add_score(scores, value):
    scores.append(value)


class_scores = [88, 91]

add_score(class_scores, 76)

print(class_scores)
````

Expected output:

```text
[88, 91, 76]
```

During the call:

```text
class_scores ─────┐
                  ├──▶ [88, 91]
scores parameter ─┘
```

The caller name `class_scores` and the parameter name `scores` refer to the same list.

Then:

```text
scores.append(76)
        ↓
shared list changes
        ↓
class_scores sees [88, 91, 76]
```

A function receives a reference to the list used as the argument. It does not automatically receive a separate copy.

````python
def add_two(values):
    values.append(2)


numbers = [0, 1]
add_two(numbers)

print(numbers)
````

# 7. References and aliases

A **reference** is the connection from a name to an object.

Assignment can create two names that refer to the same list:


````python
a = [1, 2, 3]
b = a

print(a)
print(b)
````

Reference model:

```text
a ───┐
     ├──▶ [1, 2, 3]
b ───┘
```

When two names refer to the same object, the names are **aliases** for that object.


````python
a = [1, 2, 3]
b = a

b.append(4)

print(a)
print(b)
````

Both names show:

```text
[1, 2, 3, 4]
```

The program mutated one shared list object.


## Reassigning one alias

The names are separate names. One name can later be assigned to a different list.


````python
a = [1, 2, 3]
b = a

b = [9]

print(a)
print(b)
````

Expected:

```text
[1, 2, 3]
[9]
```

After the reassignment:

```text
a ─────▶ [1, 2, 3]

b ─────▶ [9]
```


# 8. Copying a list

Sometimes a program needs equal starting contents and independent future changes.

`copy()` creates a separate list with the same current contents.


````python
a = [1, 2, 3]
b = a.copy()

b.append(4)

print(a)
print(b)
````

Expected:

```text
[1, 2, 3]
[1, 2, 3, 4]
```

Visual:

```text
a ─────▶ [1, 2, 3]

b ─────▶ [1, 2, 3]
```

There are two separate lists.

Practical rule:

```text
b = a
→ shared list

b = a.copy()
→ separate list with the same current contents
```

# Vocabulary

**list** — an ordered sequence of values that can be changed  
**mutable** — able to be changed in place  
**mutation** — a change to an existing object  
**append** — add one value to the end of a list  
**reference** — the connection from a name to an object  
**alias** — another name referring to the same object  
**copy** — a separate object created with the same current contents  
**item assignment** — replacing the value at a list index
