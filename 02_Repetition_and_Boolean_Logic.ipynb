# 2. Tracing repeated execution

A loop repeats every indented statement in its body.


````python
for number in range(3):
    print(number)
    print("Checking...")
````

Trace the loop one iteration at a time:

| Iteration | `number` | First body line | Second body line |
|---:|---:|---|---|
| 1 | `0` | print `0` | print `Checking...` |
| 2 | `1` | print `1` | print `Checking...` |
| 3 | `2` | print `2` | print `Checking...` |

```text
get next value
      ↓
number refers to it
      ↓
run every indented line
      ↓
go back for next value
```

The loop stops after there are no more values from the range.

# 3. Keeping a running total

A loop can repeat an action. Many programs also need to remember what happened during earlier iterations.

An **accumulator** is a variable that keeps a running result across iterations.


````python
total = 0

for number in range(1, 4):
    total = total + number

print(total)
````

Trace the changing state:

| Iteration | `number` | `total` before | Calculation | `total` after |
|---:|---:|---:|---|---:|
| 1 | `1` | `0` | `0 + 1` | `1` |
| 2 | `2` | `1` | `1 + 2` | `3` |
| 3 | `3` | `3` | `3 + 3` | `6` |

`total = 0` runs once before the loop begins. Each iteration then updates the same running total.

# 4. Counting when an event occurs

A **counter** records how many times an event occurs.

The counter below increases only when the current number is greater than `3`.


````python
count = 0

for number in range(1, 6):
    if number > 3:
        count = count + 1

print(count)
````

Trace the event:

| `number` | `number > 3` | `count` after |
|---:|---|---:|
| `1` | `False` | `0` |
| `2` | `False` | `0` |
| `3` | `False` | `0` |
| `4` | `True` | `1` |
| `5` | `True` | `2` |

The counter stays unchanged when the condition is `False`.


## A useful remainder operator

The `%` operator gives the remainder after division.

```text
10 % 3 → 1
```

A number is even when dividing by `2` leaves a remainder of `0`.


````python
for number in range(1, 6):
    if number % 2 == 0:
        print(number)
````

# 5. Combining conditions

In Lesson 1, comparisons produced `True` or `False`.

Boolean logic lets a program combine or reverse those conditions.


````python
fuel = 70
destination = 1

print(fuel >= 30 and destination == 1)
````

`and` is `True` when both conditions are `True`.

The next example checks whether a destination is one of the two Mission Control choices.


````python
destination = 2

print(destination == 1 or destination == 2)
````

`or` is `True` when at least one condition is `True`.

`not` reverses a Boolean value.


````python
fuel = 70
ready = fuel >= 30

print(ready)
print(not ready)
````

# 6. Repeating while a condition is true

A `for` loop works well when a program has a known count or set of values to step through.

A `while` loop repeats while its condition remains `True`.


````python
battery = 20

while battery < 80:
    battery = battery + 20
    print(battery)
````

Trace the condition and update:

| Check | `battery < 80` | Update | Printed |
|---:|---|---|---:|
| 1 | `20 < 80` → `True` | `20 + 20` | `40` |
| 2 | `40 < 80` → `True` | `40 + 20` | `60` |
| 3 | `60 < 80` → `True` | `60 + 20` | `80` |
| 4 | `80 < 80` → `False` | stop | — |

```text
check condition
      ↓ True
run body
      ↓
update state
      ↓
check condition again

False → stop
```

The line `battery = battery + 20` eventually makes the condition `False`.


## A `while` loop with user input

The changing state can come from user input.


````python
savings = 0

while savings < 100:
    deposit = int(input("Deposit amount: "))
    savings = savings + deposit

print("Goal reached!")
````

Use this as a beginner design guide. Python allows more flexible choices as programs become more advanced.

# 8. Chapter application — Reading Analyzer

The Reading Analyzer repeats the same input process several times and keeps two pieces of changing state:

- a running total,
- a count of alert readings.

A reading is an alert when it is below `20` **or** above `80`.

## Program flow

```text
ask how many readings
        ↓
repeat that many times
        ↓
receive one reading
        ↓
add it to total
        ↓
check alert condition
        ↓
update alert count when needed
        ↓
print summary
```

Complete TODO 1–6.


````python
number_of_readings = int(input("How many readings? "))

# TODO 1 — running total starts at zero
total = -1

# TODO 2 — alert counter starts at zero
alert_count = -1

# TODO 3 — repeat once for each reading
for number in range(0):
    reading = int(input("Enter a reading: "))

    # TODO 4 — add the reading to the running total
    total = total + 0

    # TODO 5 — alert when the reading is below 20 or above 80
    if reading < 0 or reading > 0:

        # TODO 6 — increase the alert counter by 1
        alert_count = alert_count + 0

print(f"Total: {total}")
print(f"Alert readings: {alert_count}")
````

# Vocabulary

**loop** — a structure that repeats a block of code  
**iteration** — one execution of a loop body  
**loop variable** — the variable that receives the current value during a `for` loop  
**range** — a built-in that provides integer values for controlled repetition  
**accumulator** — a variable that keeps a running result  
**counter** — a variable that records how many times an event occurs  
**Boolean logic** — operations that combine or reverse `True`/`False` conditions  
**while loop** — repetition that continues while a condition is `True`
