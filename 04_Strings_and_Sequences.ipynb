# 1. Indexing

An **index** is a position in a sequence.

Python starts string indexes at `0`.


````python
word = "PYTHON"

print(word[0])
print(word[3])
````

The output is:

```text
P
H
```

Trace the second lookup:

```text
word[3]
   ↓
index 3
   ↓
"H"
```

# 2. Measuring a string with `len()`

`len()` returns the number of values in a sequence.

For a string, it returns the number of characters.


````python
word = "PYTHON"
length = len(word)

print(length)
````

The output is:

```text
6
```

Connect length to valid indexes:

```text
length = 6
valid indexes = 0 through 5
last index = length - 1
```

A program can calculate the last index:


````python
word = "PYTHON"

last_index = len(word) - 1
print(word[last_index])
````

# 3. Slicing

One index selects one character.

A **slice** extracts a section of a sequence.


````python
word = "PYTHON"
print(word[0:3])
````

The output is:

```text
PYT
```

The pattern is:

```text
[start : stop]
```

The start position is included. The stop position is excluded.

This is the same boundary idea used by:

```python
range(start, stop)
```

Visual:

```text
index   0   1   2   3   4   5
        P   Y   T   H   O   N
        └───────┘
        word[0:3]
           "PYT"
```

# 4. Looping through characters

A `for` loop can step through the characters in a string one at a time.


````python
word = "CODE"

for char in word:
    print(char)
````

Trace the loop:

| Iteration | `char` |
|---:|---|
| 1 | `"C"` |
| 2 | `"O"` |
| 3 | `"D"` |
| 4 | `"E"` |

The pattern is:

```text
string
  ↓
next character
  ↓
loop variable
  ↓
body
```

The loop variable holds the current character, not the current index.


## Reusing the counter pattern

Lesson 2 introduced counters. The same pattern works while looping through text.


````python
count = 0

for char in "BANANA":
    if char == "A":
        count = count + 1

print(count)
````

Expected output:

```text
3
```

# 5. Membership with `in`

The word `in` can also ask whether a value appears inside a sequence.


````python
word = "PYTHON"
print("Y" in word)
````

Expected:

```text
True
```

The two uses of `in` have related but different jobs:

```text
for char in word
→ step through values

"Y" in word
→ ask whether a value appears
```

Membership produces a Boolean value.


````python
message = "robot ready"
print("ready" in message)
````

Membership can also be used directly in a decision:


````python
message = "system ready"

if "ready" in message:
    print("Start")
````

# 6. String methods

You already know function-call syntax:

```python
len(word)
```

Strings also provide operations written after the string value.

A **method** is an operation written after the value it operates on.

The basic pattern is:

```text
value.method()
```


````python
name = "Maya"
print(name.lower())
````

`lower()` is a method attached to a string value.

Compare:

```text
len(word)
→ function call

word.lower()
→ method call
```

Both use parentheses to run an operation.


## Common string methods

### `lower()`

Creates a lowercase string:


````python
message = "Ready"
clean = message.lower()

print(clean)
````

### `strip()`

Removes whitespace from the beginning and end:


````python
message = "  ready  "
clean = message.strip()

print(clean)
````

### `replace()`

Creates a new string with specified text replaced:


````python
message = "robot ready"
updated = message.replace("robot", "system")

print(updated)
````

### `split()` — preview

`split()` separates text into pieces:


````python
sentence = "red blue green"
print(sentence.split())
````

Python stores the pieces together in a collection.

Lesson 5 studies that collection structure in detail. For now, the important idea is what the text method does. This lesson does not require indexing or modifying the result of `split()`.


# 7. Strings are immutable

Strings are **immutable**.

After a string is created, its characters cannot be changed in place.


````python
message = "Ready"
clean = message.lower()

print(message)
print(clean)
````

The output is:

```text
Ready
ready
```

Mental model:

```text
message ─▶ "Ready"

message.lower()
      ↓
new string
      ↓
"ready"

clean ─▶ "ready"
message remains "Ready"
```

The method creates a new string. The original string stays unchanged.


## Reassignment

A variable can be assigned the new string:


````python
message = "Ready"
message = message.lower()

print(message)
````

This assigns the new string produced by `lower()` to the name `message`.

It does not edit the original string in place.

A direct character assignment would not be valid:

```python
word = "CAT"
# word[0] = "B"
```

Individual string positions cannot be assigned a new character.


# Debugging: index outside the string

Consider:


````python
word = "CODE"

# The line below would try to access a position outside the string:
# print(word[4])

print(word[3])
````

Reason through the indexes:

```text
len(word) = 4
valid indexes = 0, 1, 2, 3
```

So index `4` is outside the string.

If the goal is always the final character, this version adapts to the string length:


````python
word = "CODE"
print(word[len(word) - 1])
````
