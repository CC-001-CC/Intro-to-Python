# 3. Object state

**State** is the current data that describes an object at a particular moment.

For mutable objects, operations can change that state.


````python
scores = [88, 91]

print(scores)

scores.append(95)

print(scores)
````

Trace:

```text
before
scores ─▶ [88, 91]

append(95)

after
scores ─▶ [88, 91, 95]
```

The list is the same mutable object with updated state.


````python
device = {
    "name": "Sensor A",
    "battery": 80,
    "status": "ready"
}

print(device)

device["battery"] = 65

print(device)
````

For the device record:

```text
before
battery → 80

after
battery → 65
```

The dictionary's current fields and values represent its state.


## Immutable objects still have values and operations

Strings are objects too:


````python
name = "maya"
upper_name = name.upper()

print(name)
print(upper_name)
````

Expected:

```text
maya
MAYA
```

The original string value remains unchanged. The method call produces another string value.

A useful summary is:

```text
mutable list / dictionary
→ an operation may change existing state

immutable string
→ string operations produce another string value
```


# 4. References connect names to objects

Lesson 5 introduced references. The same model applies to dictionary records.


````python
device = {
    "name": "Sensor A",
    "battery": 80
}

current = device

current["battery"] = 60

print(device)
print(current)
````

Reference model:

```text
device ───┐
          ├──▶ {"name":"Sensor A", "battery":60}
current ──┘
```

A **reference** connects a name to an object.

The program above has two names referring to one dictionary object. A mutation through either name changes the shared object.


## Function parameters use the same reference model

When a mutable object is supplied as an argument, the caller's name and the parameter can refer to the same object during the call:

```text
caller name ─┐
             ├──▶ one mutable object
parameter ───┘
```

This explains why a function can change state that remains visible after the function call finishes.


# 5. Function interfaces

A **function interface** describes three things:

1. what the function needs,
2. what it returns,
3. what it may change.

Consider a function that reads device state.


````python
def is_low_battery(device):
    return device["battery"] < 30


device = {
    "name": "Sensor A",
    "battery": 20
}

print(is_low_battery(device))
````

Interface:

| Question | Answer |
|---|---|
| Needs | one device dictionary |
| Returns | `True` or `False` |
| May change | nothing |

The function reads state and returns information.


````python
def set_status(device, status):
    device["status"] = status


device = {
    "name": "Sensor A",
    "status": "ready"
}

set_status(device, "charging")

print(device)
````

Interface:

| Question | Answer |
|---|---|
| Needs | a device dictionary and a status |
| Returns | `None` |
| May change | the device's `"status"` field |

The general model is:

```text
needs
  ↓
function
  ├─→ returns
  └─→ may change state
```


### READ

For each function, identify:

```text
needs
returns
may change
```

This is an interface description. It focuses on how other code can use the function.


# 6. Returning information and changing state

A function can read state and return information:


````python
def is_low_battery(device):
    return device["battery"] < 30


device = {
    "name": "Sensor A",
    "battery": 20
}

result = is_low_battery(device)

print(result)
print(device)
````

The Boolean result is information about the device. The function leaves the device dictionary unchanged.

A different function can intentionally mutate state:


````python
def charge_device(device, amount):
    device["battery"] = device["battery"] + amount


device = {
    "name": "Sensor A",
    "battery": 20
}

charge_device(device, 15)

print(device["battery"])
````

Expected:

```text
35
```

During the call:

```text
caller device ───┐
                 ├──▶ {"battery":20}
parameter device ┘
```

Then the function updates the shared dictionary:

```text
battery 20
    ↓
battery 35
```

A **side effect** is an observable change caused by an operation, such as changing a mutable object's state.


# Debugging: a returned value does not automatically change state

Read this function:


````python
def battery_after_charge(device, amount):
    return device["battery"] + amount


device = {
    "name": "Sensor A",
    "battery": 20
}

battery_after_charge(device, 15)

print(device["battery"])
````

Expected:

```text
20
```

Trace:

```text
function computes 35
       ↓
returns 35
       ↓
caller leaves the returned value unused
       ↓
device battery remains 20
```

The caller can use the returned value:


````python
def battery_after_charge(device, amount):
    return device["battery"] + amount


device = {
    "name": "Sensor A",
    "battery": 20
}

new_battery = battery_after_charge(device, 15)

print(new_battery)
print(device["battery"])
````

# 8. The data + behavior design problem

The Device Toolkit now contains structured device data:

```text
DEVICE DATA
name
battery
status
```

and a related group of operations:

```text
DEVICE OPERATIONS
is_low_battery(...)
charge_device(...)
set_status(...)
show_device(...)
```

Each operation accepts a device because it works with device state.

When a program has many device records, the same operations apply to each device.

This creates a design question:

> The data describes a device, and these functions operate on a device. How could Python let us define our own type that keeps these ideas together?

# Vocabulary

**object** — a Python value with a type and associated operations  
**type** — the category that helps determine an object's behavior  
**method** — an operation associated with an object  
**state** — current data describing an object  
**reference** — a connection from a name to an object  
**interface** — what a function needs, returns, and may change  
**side effect** — an observable change caused by an operation  
**behavior** — operations that act on or use data
